```python
# Feature Engineering

# Lag features: previous prices
df["Price_Lag_1"] = df["Price"].shift(1)
df["Price_Lag_7"] = df["Price"].shift(7)
df["Price_Lag_30"] = df["Price"].shift(30)

# Moving averages
df["MA_7"] = df["Price"].rolling(window=7).mean()
df["MA_30"] = df["Price"].rolling(window=30).mean()
df["MA_90"] = df["Price"].rolling(window=90).mean()

# Momentum: change in price over 7 days
df["Momentum_7"] = df["Price"] - df["Price_Lag_7"]

# 30-day rolling volatility
df["Daily_Return"] = df["Price"].pct_change()
df["Volatility_30"] = df["Daily_Return"].rolling(window=30).std()

print(df[[
    "Date",
    "Price",
    "Price_Lag_1",
    "Price_Lag_7",
    "Price_Lag_30",
    "MA_7",
    "MA_30",
    "MA_90",
    "Momentum_7",
    "Volatility_30"
]].tail())
```

```python
# Remove rows with missing values created by lag features

df = df.dropna().reset_index(drop=True)

print("Dataset shape after feature engineering:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Create SQLite database for GoldSense

import sqlite3

# Create/connect to the database
conn = sqlite3.connect("goldsense.db")

# Store the feature-engineered dataset in a SQL table
df.to_sql("gold_prices", conn, if_exists="replace", index=False)

print("GoldSense database created successfully.")
```

# Load dataset

```python
df = pd.read_csv("../data/datasetfull.csv", header=None)

print(df.shape)
print(df.head())
print(df.tail())
print(df.info())
```

## Give the columns proper names

```python
df.columns = ["Date", "Price"]

print(df.head())
print(df.dtypes)
```

## Convert Date

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

print(df.dtypes)
print(df.head())
```

## Sort the data by date

```python
df = df.sort_values("Date").reset_index(drop=True)

print(df.head())
print(df.tail())
```

## Check data quality

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nPrice <= 0:", (df["Price"] <= 0).sum())
```

## Check the date range and basic price statistics

```python
print("Date range:", df["Date"].min(), "to", df["Date"].max())

print("\nPrice statistics:")
print(df["Price"].describe())
```

## First EDA: Overall Gold Price Trend

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["Price"])

plt.title("Gold Price Trend (2011–2018)")
plt.xlabel("Date")
plt.ylabel("Gold Price")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Daily percentage change in gold price
df["Daily_Return"] = df["Price"].pct_change() * 100

print(df[["Date", "Price", "Daily_Return"]].head(10))
```

### Moving Averages

```python
# Calculate moving averages to understand short-term and long-term trends
df["MA_7"] = df["Price"].rolling(window=7).mean()
df["MA_30"] = df["Price"].rolling(window=30).mean()

print(df[["Date", "Price", "MA_7", "MA_30"]].head(35))
```

### Plot Moving Averages

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["Price"], label="Gold Price")
plt.plot(df["Date"], df["MA_7"], label="7-Day Moving Average")
plt.plot(df["Date"], df["MA_30"], label="30-Day Moving Average")

plt.title("Gold Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Gold Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 30-Day Volatility

```python
# Calculate 30-day rolling volatility
df["Volatility_30"] = df["Daily_Return"].rolling(window=30).std()

print(df[["Date", "Daily_Return", "Volatility_30"]].tail(10))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["Volatility_30"])

plt.title("30-Day Gold Price Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility (%)")

plt.grid(True)
plt.tight_layout()
plt.show()
```

## Anomaly Detection

### Detect unusual price movements using IQR

```python
# Detect unusual gold price changes using the IQR method

Q1 = df["Daily_Return"].quantile(0.25)
Q3 = df["Daily_Return"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

### Mark the anomalies

```python
df["Is_Anomaly"] = (
    (df["Daily_Return"] < lower_limit) |
    (df["Daily_Return"] > upper_limit)
)

print("Number of anomalies:", df["Is_Anomaly"].sum())
```

### See the actual anomalous dates

```python
anomalies = df[df["Is_Anomaly"] == True]

print(anomalies[["Date", "Price", "Daily_Return"]].head(20))
```

## yearly average price

```python
# Calculate yearly average gold price

df["Year"] = df["Date"].dt.year

yearly_avg = df.groupby("Year")["Price"].mean()

print(yearly_avg)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(yearly_avg.index, yearly_avg.values)

plt.title("Average Gold Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Gold Price")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Monthly Average Gold Price

```python
# Calculate monthly average gold price

df["Month"] = df["Date"].dt.month

monthly_avg = df.groupby("Month")["Price"].mean()

print(monthly_avg)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(monthly_avg.index, monthly_avg.values, marker="o")

plt.title("Average Gold Price by Month")
plt.xlabel("Month")
plt.ylabel("Average Gold Price")

plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

### High–Low Price Range

```python
# Calculate daily price range

df["Price_Range"] = df["Price"].max() - df["Price"].min()

print("Overall Gold Price Range:", df["Price_Range"].iloc[0])
```
